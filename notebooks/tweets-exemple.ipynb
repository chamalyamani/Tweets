```python
def load_tweets(file_path, sentiment):
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()

    tweets = [tweet.strip() for tweet in content.split(",") if tweet.strip()]

    df = pd.DataFrame({
        "tweet": tweets,
        "sentiment": sentiment
    })

    return df
```
